# EPL exploratory analysis — reproducibility notebook

The historical data is in my original GitHub repository (`results.csv` and `stats (1).csv`). This version makes input checks explicit instead of repeating old screenshots or making up rankings.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
folder = Path('data')
results_path = folder/'results.csv'
stats_path = folder/'stats (1).csv'
print('results exists?', results_path.exists(), 'stats exists?', stats_path.exists())
```

```python
if not results_path.exists():
    print('Download results.csv from the original GitHub repo first. No statistics computed.')
else:
    games = pd.read_csv(results_path)
    expected = {'home_team','away_team','home_goals','away_goals','result','season'}
    assert expected.issubset(games.columns), f'Missing columns: {expected-set(games.columns)}'
    display(games.head())
    print('rows',len(games),'seasons',games['season'].nunique())
    print('null counts:')
    display(games.isna().sum().to_frame('nulls'))
    home_wins = games.loc[games['result']=='H','home_team'].value_counts()
    away_wins = games.loc[games['result']=='A','away_team'].value_counts()
    total_wins = home_wins.add(away_wins,fill_value=0).sort_values(ascending=False)
    print('Top teams by total wins across available seasons:')
    display(total_wins.head(6).rename('total_wins'))
    print('Note: total wins is affected by number of seasons; do not call this best-team ranking without normalizing.')
```
