# Sentiment analysis — auditing an impressive old metric

Original notebook prints **0.9950** accuracy on 2,000 NLTK tweets. That is an observed historical output, **not an independent new evaluation**. I want to inspect how the 8k/2k split was constructed and test weaknesses.

```python
from pathlib import Path
import nltk
from nltk.corpus import twitter_samples
try:
    pos = twitter_samples.strings('positive_tweets.json')
    neg = twitter_samples.strings('negative_tweets.json')
    print('Available:', len(pos), 'positive and', len(neg), 'negative tweets')
except LookupError:
    print('NLTK twitter_samples corpus missing. In an environment with permitted downloads: nltk.download("twitter_samples").')
    pos = neg = []
```

## The split and the leakage question

The original code uses the first 4,000 positive and first 4,000 negative tweets for training, and the remaining 1,000 of each for test. Its `(word,label)` frequency features are computed from training tweets. This avoids the most obvious split leak, but temporal/domain generalization is still unknown.

```python
if pos and neg:
    train_x = pos[:4000] + neg[:4000]
    test_x = pos[4000:] + neg[4000:]
    print('train', len(train_x), 'test', len(test_x))
    print('overlapping literal tweets', len(set(train_x).intersection(test_x)))
else:
    print('Dataset unavailable; do not manufacture a re-evaluation.')
```
